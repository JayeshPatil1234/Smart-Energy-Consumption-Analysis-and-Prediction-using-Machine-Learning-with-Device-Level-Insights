```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/smart_home_energy_consumption_large.csv")

print(df.columns)
df.head()
```

```python
df.columns = df.columns.str.strip()
```

```python
df["timestamp"] = pd.to_datetime(
    df["Date"].astype(str) + " " + df["Time"].astype(str),
    errors="coerce"
)

df = df.sort_values("timestamp")
df[["Date", "Time", "timestamp"]].head()
```

```python
df["hour"] = df["timestamp"].dt.hour
df["day"] = df["timestamp"].dt.day
df["weekday"] = df["timestamp"].dt.weekday
df["month"] = df["timestamp"].dt.month

df[["timestamp", "hour", "weekday", "month"]].head()
```

```python
df["lag_1"] = df["Energy Consumption (kWh)"].shift(1)
df["lag_24"] = df["Energy Consumption (kWh)"].shift(24)
```

```python
df["rolling_mean_24"] = (
    df["Energy Consumption (kWh)"]
    .rolling(window=24)
    .mean()
)
```

```python
df = pd.get_dummies(
    df,
    columns=["Appliance Type", "Season"],
    drop_first=True
)
```

```python
df = df.drop(columns=["Date", "Time"])
df = df.dropna()

df.to_csv("feature_engineered_energy.csv", index=False)

df.shape
```
