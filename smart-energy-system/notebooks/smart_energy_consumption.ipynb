```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("smart_home_energy_consumption_large.csv")

df["Date"] = pd.to_datetime(df["Date"])
df.set_index("Date", inplace=True)
```

```python
weekly_energy = df["Energy Consumption (kWh)"].resample("W").sum()

plt.figure()
plt.plot(weekly_energy)
plt.title("Weekly Energy Consumption")
plt.xlabel("Week")
plt.ylabel("Total Energy Consumption (kWh)")
plt.show()
```

```python
monthly_energy = df["Energy Consumption (kWh)"].resample("M").sum()

plt.figure()
plt.plot(monthly_energy)
plt.title("Monthly Energy Consumption")
plt.xlabel("Month")
plt.ylabel("Total Energy Consumption (kWh)")
plt.show()
```

```python
monthly_energy.plot(kind="bar")
plt.title("Monthly Energy Consumption")
plt.xlabel("Month")
plt.ylabel("Energy Consumption (kWh)")
plt.show()
```
